```python
# Bereken hier je benodigde kantelfrequentie
#1/3 demping geeft 1/2omega^2 = omega_0^2
w0 = ((0.25*(2*np.pi*5)**2)**0.5)
print(w0)
f0 = w0/(2*np.pi)
print(f0)
```

```python
# Voeg hier je plot in.
t = np.linspace(1,70, 1000)
def verhouding(w, w0):
    return 1/(1-((w**2)/(w0**2)))
plt.plot(t, abs(verhouding(t, w0)))
plt.xscale('log')
plt.yscale('log')
plt.plot()
```

```python
# Bereken hier de minimale Delta-x waarbij je aan de eis van de opdracht voldoet.
dx = (9.81/(w0**2))
print(dx)
```
