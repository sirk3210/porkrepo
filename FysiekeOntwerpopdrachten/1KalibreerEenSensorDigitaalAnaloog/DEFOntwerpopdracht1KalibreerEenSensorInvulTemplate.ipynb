## *Opdracht 6*: Meten en fitten, Leerdoel

```python
# Zet hier jouw data in een voor jou handig formaat
import numpy as np

ductape: 896+1
telefoon: 670+3

massa_meter = np.array([766, 872, 599, 520, 751, 781, 807])
u_massa_meter = np.array([1, 1, 1 ,3, 1, 1, 1])
massa_ref = np.array([237.71, 436.68, 100.72, 84.58, 150.83, 214.43, 115.9])+67.7
# telefoon, dutape, rekenmachine, plank 1, plank1+2, plank 1+2+3, 4+2
# massa_meter = np.array([745,  644, 395, 433])
# u_massa_meter = np.array([10, 1, 5, 5])
# massa_ref = np.array([438.68,  237.31, 100.72, 342.5])+67.7
u_massa_ref = np.ones(7)*0.1
```

```python
# Plot hier de meting en fit een lijn door de metingen.
# Maak gebruik van wat je geleerd hebt bij het practicum.
%matplotlib inline
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

def f (x, a, b):
    return a*x+b

# def f(x,a):
#     return a*x**2

massa_meter1 = np.sort(massa_meter)
massa_ref1 = np.sort(massa_ref)


values, covariance = curve_fit(f, massa_meter1, massa_ref1)

t = np.linspace(min(massa_meter), max(massa_meter), 1000)


plt.plot(t, f(t, values[0], values[1]))
# plt.plot(t, f(t, values[0]))
plt.errorbar(massa_meter, massa_ref, u_massa_meter, u_massa_ref, '+')
plt.show
```
